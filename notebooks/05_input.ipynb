````python
=# Ejemplito
# El usuario provee los 3 lados de un triangulo, con esta informacion
# proceda a calcular el area del triangulo

# Entradas/Inputs
lado1 = float(input("Ingresa el primer lado del triangulo  : "))
lado2 = float(input("Ingresa el segundo lado del triangulo  : "))
lado3 = float(input("Ingresa el tercer lado del triangulo  : "))

# Basandome en un resultado EMINENTEMENTE TEORICO procedo con la obtencion
# del area del triangulo
# Utilizando la Formula de Heron
semiperimetro = (lado1 + lado2 + lado3)/2
area = (semiperimetro * (semiperimetro-lado1) * (semiperimetro - lado2) * (semiperimetro -lado3))**(0.5)

# Mostremos el valor de la variable area
area
````

````python
lado_A = float(input("Ingrese la longitud del lado A: "))
lado_B = float(input("Ingrese la longitud del lado B: "))
lado_C = float(input("Ingrese la longitud del lado C: "))

semiperimetro = 0.5 * (lado_A + lado_B + lado_C)

dif_A = semiperimetro - lado_A
dif_B = semiperimetro - lado_B
dif_C = semiperimetro - lado_C

producto = semiperimetro * dif_A * dif_B * dif_C

area = producto ** 0.5

print(area)
````

````python
# Ejemplito
# Dado el valor de un angulo x , aproximar mediante el polinomio de taylor la funcion
# seno(x)  approx x - x**3/3! + x**5/5! - x**7/7!

# Entradas/Inputs
angulo = float(input("Ingresa el angulo (°): "))

# Basandome en un resultado EMINENTEMENTE TEORICO procedo a obtener una aproximacion
# del seno del angulo ingresado (proveido por el usuario)

# Necesitamos importar el modulo math
import math as m

# Pasamos/transformamos el angulo a radianes
x = angulo * m.pi / 180

# Utilicemos la formula de aproximacion
seno_taylor = x - (x**3)/(m.factorial(3)) + (x**5)/(m.factorial(5)) - (x**7)/(m.factorial(7))

# Mostremos el resultado
seno_taylor
````

````python
# Calculemos el seno(30°) usando el modulo math
m.sin(m.radians(30))
````

````python
# Calculemos el error cometido por nuestra implementacion (Valor teorico : 0.5)
error1 = m.fabs(0.5 - seno_taylor)
error1
````

````python
# Calculemos el error que comete el modulo math
error2 = m.fabs(0.5 - m.sin(m.radians(30)))
error2
````

````python
# En vista que el error cometido por el modulo math es menor que el
# error cometido por mi implementacion , concluimos que la aproximacion
# que utiliza el modulo math es mas precisa
````

````python
# Necesito mostrar en una misma celda las variables error1 y error2
error1
error2
````

````python
# Ejemplito
# Calculemos el interes simple de un monto proveido por el usuario (M0) a una
# tasa (i) del 10% anual con un horizonte (t) dado por el usuario

# Inputs
M0 = float(input("Ingresa el monto inicial : "))
t = float(input("Inigresa el tiempo (resolucion anual) : "))

# Operaciones (Teorico)
i = 0.1
interes = M0 * i * t

# Salida/Outputs
print("El interes generado es: " , interes)
print("El interes generado es %.2f" %(interes))
````

````python
# Ejemplito
# Consideremos pedir al usuario su edad (almacenada en memoria ram como un int)
# En funcion del valore ingresado, tomaremos algunas acciones :
    # edad >= 0 y edad<18 : "Es momento de preparar el terreno para la siembra"
    # edad >=18 y edad<29 : "Es momento de sembrar"
    # edad >= 29 y edad < 45 : "Es momento de cuidar el terreno para futuras siembras"
    # edad >= 45 y edad < 99 : "Es momento de disfrutar la cosecha"

print("Bienvenidos a mi primera celda que utiliza la estructura de decision IF")

# Entradas/Inputs
edad_user = int(input("Ingresa tu edad (Numero Entero) : "))

# Implementemos la logica
if (edad_user >= 0 ) and (edad_user < 18):
    print("Es momento de preparar el terreno para la siembra")
elif (edad_user >= 18) and (edad_user < 29):
    print("Es momento de sembrar")
elif (edad_user >= 29) and (edad_user < 45):
    print("Es momento de cuidar el terreno para futuras siembras")
elif (edad_user >= 45) and (edad_user < 99) :
    print("Es momento de disfrutar la cosecha")
else:
    print("El valor ingresado no es una edad adecuada (considerada)")
````
