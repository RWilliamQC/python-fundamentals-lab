```python
# Pidamos a python que nos comunique el espacio en memoria que ocupan las variables
import sys
sys.getsizeof(x1)
```

```python
# Este numero 10**10 ocupa 32B
sys.getsizeof(10**10)
```

```python
# El numero 10**100 ocupa 72B
sys.getsizeof(10**100)
```

```python
# Este numero mas pequeño que x1 ocupa 24B
sys.getsizeof(10**(-1*x1))
```
