## Operaciones con variables de naturaleza numerica

- `+` : operacion suma
- `-` : operacion resta
- `*` : multiplicacion
- `/` : division
- `**` : potenciacion

**Observacion** : No hay que olvidar el uso de parentesis para operaciones complejas.

Algoritmo de Euclides : `D = d * q + r`

    * Cociente : `q = D // d`

    * Residuo : `r = D % d`


```python
# Ejemplito : Un rectangulo tiene un perimetro de 34cm y un area de 60cm2. Halla sus dimensiones
# y luego construye un cuadrado con el mismo perimetro y calcula su area.
# Tambien puedes calcular la diferencia de areas.
```

```python
# Para el rectangulo :
    # perimetro = 2 * base + 2 * altura = 34
    # area = base * altura = 60
# De la primera ecuacion : base = (34 - 2*altura)/2 = 17 - altura
# En la segunda ecuacion : (17-altura) * altura = 60
# Operando obtenemos las dimensiones del rectangulo
    # base = 12
    # altura = 5
#
# Cuadrado con el mismo perimetro : 4 * lado = 34
lado = 8.5
#
# Calculo del area
area = lado**2
area
```

```python
# Diferencia de areas
area - 60
```
